# Plotting HandZ vs. Time: A Walkthrough

This notebook demonstrates a standard approach to loading and visualizing time-series data in Python. Here's a breakdown of the tools and the design intuition behind them.

### 1. The Environment: Jupyter Notebook

-   **Why a notebook?** A Jupyter Notebook provides an interactive environment ideal for data exploration. We can write and execute code in small, manageable chunks called "cells."
-   **Immediate Feedback:** This lets us see the output of each step immediately. For example, after loading the data, we can instantly display the first few rows to verify it's correct. When we create a plot, it appears right there in the notebook. This iterative process is incredibly efficient.

### 2. Data Handling: The `pandas` Library

-   **Why `pandas`?** The `pandas` library is the industry standard for data manipulation in Python. It's built to handle structured data (like the rows and columns in your CSV file) efficiently.
-   **The DataFrame:** The core of pandas is the DataFrame. You can think of it as a smart spreadsheet inside your code. It makes tasks that are complex in standard Python (like selecting columns or filtering rows) very simple.
-   **`pd.read_csv()`:** This function is a powerful and fast way to import data directly from a CSV file into a DataFrame. It's the most direct path from your file to a usable data structure.

### 3. Plotting: The `matplotlib` Library

-   **Why `matplotlib`?** It is the foundational plotting library in Python. It gives you complete control over every element of your plot.
-   **The Design of the Plot:**
    -   **Line Plot:** A line plot (`plt.plot()`) is the best way to visualize how a continuous value (like `HandZ`) changes over time. It connects the data points to clearly show trends, peaks, and valleys.
    -   **Figure and Axes:** Creating a `figure` (`plt.figure()`) is like getting a blank canvas. It allows us to set properties for the whole plot, such as its size (`figsize`).
    -   **Labels and Title:** A plot without labels is ambiguous. Adding a `title`, `xlabel`, and `ylabel` is critical for making the graph understandable.
    -   **Grid and Legend:** The `grid` helps in tracing points back to the axes to estimate values, and the `legend` is essential if you plot multiple variables on the same graph.

```python
import pandas as pd

# Load the dataset
file_path = 'TA_OCULUS_PILOT_JUL4_2_OCULUS_interpolated.csv'
data = pd.read_csv(file_path)

# Define the columns to rename
columns_to_rename = {
    'TransformPosX': 'HandX',
    'TransformPosY': 'HandY',
    'AdditionZ': 'HandZ'
}

# Rename the columns
data.rename(columns=columns_to_rename, inplace=True)

# Display the first few rows to confirm the new column names
data.head()
```

```python
import matplotlib.pyplot as plt

# Ensure the 'Timestamp' column is in datetime format for better plotting
data['Timestamp'] = pd.to_datetime(data['Timestamp'])

# Create a figure and axis for the plot
plt.figure(figsize=(15, 7))

# Plot the HandZ data against the timestamp
plt.plot(data['Timestamp'], data['HandZ'], label='HandZ')

# Add titles and labels for clarity
plt.title('HandZ Position Over Time')
plt.xlabel('Time')
plt.ylabel('HandZ Position')
plt.legend()
plt.grid(True)

# Improve date formatting on the x-axis
plt.gcf().autofmt_xdate()

# Show the plot
plt.show()
```
